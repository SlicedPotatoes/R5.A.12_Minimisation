##### Recherche du minimum

```python
import time
from algo.gradientPasFixe import algo
from utils.point import Point

def f(p: Point)->float:
    x0 = p[0]
    x1 = p[1]
    x2 = p[2]
    x3 = p[3]

    return (x0 + 10*x1)**2 + 5*(x2 - x3)**2 + (x1 - 2*x2)**4 + 10*(x0 - x3)**4

def dp_0(p:Point):
    x0 = p[0]
    x1 = p[1]
    x2 = p[2]
    x3 = p[3]

    return 2 * (x0 + 10 * x1) + 40 * (x0 - x3)**3

def dp_1(p:Point):
    x0 = p[0]
    x1 = p[1]
    x2 = p[2]
    x3 = p[3]

    return 20 * (x0 + 10 * x1) + 4 * (x1 - 2 * x2)**3

def dp_2(p:Point):
    x0 = p[0]
    x1 = p[1]
    x2 = p[2]
    x3 = p[3]

    return 20 * (x2 - x3) - 8 * (x1 - 2 * x2)**3

def dp_3(p:Point):
    x0 = p[0]
    x1 = p[1]
    x2 = p[2]
    x3 = p[3]

    return -20 * (x2 - x3) - 40 * (x0 - x3)**3

dp = [dp_0, dp_1, dp_2, dp_3]
init_point = Point([3, -1, 0, 1])

alpha = 10**-2
epsilon = 10**-3
max_iter = 100000000

times = []
result = None
history = []

for i in range(1):
    start = time.perf_counter()
    result, history = algo(dp, init_point, alpha, epsilon, max_iter)
    end = time.perf_counter()

    times.append(end - start)

print(result)
print(f(result))
print(len(history))
print("time: ", sum(times) / len(times), "seconds")
```
