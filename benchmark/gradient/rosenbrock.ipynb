##### Recherche du minimum

```python
import time
from algo.gradientPasFixe import algo
from utils.point import Point

def f(p: Point)->float:
    x0 = p[0]
    x1 = p[1]

    return (1-x0)**2 + 100*(x1-x0**2)**2

def dp_0(p:Point):
    x0 = p[0]
    x1 = p[1]

    return 2*x0 - 2 - 400*x0*x1 + 400*x0**3

def dp_1(p:Point):
    x0 = p[0]
    x1 = p[1]

    return 200*x1 - 200*x0**2

dp = [dp_0, dp_1]
init_point = Point([-1.5, 4])

alpha = 10**-3
epsilon = 10**-1
max_iter = 10**6

times = []
result = None
history = []

for i in range(1):
    start = time.perf_counter()
    result, history = algo(dp, init_point, alpha, epsilon, max_iter)
    end = time.perf_counter()

    times.append(end - start)

print(result)
print(f(result))
print(len(history))
print("time: ", sum(times) / len(times), "seconds")
```
