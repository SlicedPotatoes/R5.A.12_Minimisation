##### Recherche du minimum

```python
import time
from algo.gradientPasFixe import algo
from utils.point import Point

def f(p: Point)->float:
    _result = 0

    for x in p:
        _result += x**4

    return _result

dp = [lambda p,i=i : 4*p[i]**3 for i in range(10)]

init_point = Point([1 for i in range(10)])

alpha = 10**-3
epsilon = 10**-1
max_iter = 10**6

times = []
result = None
history = []

for i in range(1):
    start = time.perf_counter()
    result, history = algo(dp, init_point, alpha, epsilon, max_iter)
    end = time.perf_counter()

    times.append(end - start)

print(result)
print(f(result))
print(len(history))
print("time: ", sum(times) / len(times), "seconds")
```
