```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots()

x = [2, 3, 10]
y = [11, 16, 58]

ax.plot(x, y)
ax.scatter(x, y, c="r")
ax.set_xlabel("Dimension de la fonction f")
ax.set_ylabel("Temps moyens par itération (µs)")
ax.set_title("Temps moyens d'une itération en fonction de la dimension de f")

plt.show()
```

```python
fig, ax = plt.subplots()

nbr_itr = 200000

x = [i for i in range(nbr_itr)]
y_2 = [i*11 for i in range(nbr_itr)]
y_3 = [i*16 for i in range(nbr_itr)]
y_10 = [i*58 for i in range(nbr_itr)]

ax.plot(x, y_2, label="2d")
ax.plot(x, y_4, label="4d")
ax.plot(x, y_10, label="10d")

ax.set_xlabel("Nombre d'itérations")
ax.set_ylabel("Temps d'exécution (µs)")
ax.legend()
ax.set_title("Comparaison du temps d'exécution en fonction de la dimension de f")

plt.show()
```

```python
import numpy as np
%matplotlib widget
def f(x, y):
    return x**4 + y**4

x = np.arange(-5, 5, 0.01)
y = np.arange(-5, 5, 0.01)

X, Y = np.meshgrid(x, y)

Z = f(X, Y)

ax = plt.subplot(111, projection='3d')
ax.plot_surface(X, Y, Z)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x, y)")
ax.set_title("Somme Quartique 3D")

plt.show()
```
