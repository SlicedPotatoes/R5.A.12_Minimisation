```python
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots()

x = [2, 3, 10]
y = [11, 16, 58]

ax.plot(x, y)
ax.scatter(x, y, c="r")
ax.set_xlabel("Dimension de la fonction f")
ax.set_ylabel("Temps moyens par itération (µs)")
ax.set_title("Temps moyens d'une itération en fonction de la dimension de f")

plt.show()
```

```python
fig, ax = plt.subplots()

x = ["NM_R", "G_R", "NM_P", "G_P", "NM_Q", "G_Q"]

y = [0.87, 15.3, 0.8, 7.92, 7.11, 18.71 ]
colors = ["#e8833a" if i % 2 == 0 else "#1596b4" for i in range(len(x))]

ax.bar(x, y, color=colors)
ax.set_title("Temps d'exécution moyens par fonction et méthodes")
ax.set_xlabel("Méthode_Fonction")
ax.set_ylabel("Temps moyens (ms)")

plt.show()
```

```python
import numpy as np
%matplotlib widget
def f(x, y):
    return x**4 + y**4

x = np.arange(-5, 5, 0.01)
y = np.arange(-5, 5, 0.01)

X, Y = np.meshgrid(x, y)

Z = f(X, Y)

ax, fig = plt.subplots()

ax = plt.subplot(111, projection='3d')
ax.plot_surface(X, Y, Z)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x, y)")
ax.set_title("Somme Quartique 3D")

plt.show()
```
