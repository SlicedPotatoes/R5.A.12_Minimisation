##### Recherche du minimum

```python
from algo.gradientPasFixe import algo
from utils.point import Point

def f(p: Point)->float:
    x0 = p[0]
    x1 = p[1]

    return (1-x0)**2 + 100*(x1-x0**2)**2

def dp_0(p:Point):
    x0 = p[0]
    x1 = p[1]

    return 2*x0 - 2 - 400*x0*x1 + 400*x0**3

def dp_1(p:Point):
    x0 = p[0]
    x1 = p[1]

    return 200*x1 - 200*x0**2

dp = [dp_0, dp_1]
init_point = Point([0, 0])
alpha = 0.00001
epsilon = 0
max_iter = 1000000
result, history = algo(dp, init_point, alpha, epsilon, max_iter)

print(result)
print(f(result))
print(len(history))
```
